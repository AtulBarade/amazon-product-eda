Step 1 — Import libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

Step 2 — Load dataset

```python
df = pd.read_csv(r"C:\Thiranex Internship Tasks\Amazon Product Sales & Rating — Exploratory Data Analysis\Data\amazon.csv")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

Step 4 — Missing values

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": df.isnull().mean() * 100
})

missing.sort_values(
    "Percentage",
    ascending=False
)
```

Step 5 — Duplicates

```python
print(
    "Duplicate rows:",
    df.duplicated().sum()
)
```

Step 6 — Clean the numerical columns

```python
# Discounted price
df["discounted_price"] = (
    df["discounted_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
# Actual price
df["actual_price"] = (
    df["actual_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
# Dicounted Percentege
df["discount_percentage"] = (
    df["discount_percentage"]
    .str.replace("%", "", regex=False)
    .astype(float)
)
```

```python
# Rating
df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)
```

```python
# Rating Count
df["rating_count"] = (
    df["rating_count"]
    .str.replace(",", "", regex=False)
)

df["rating_count"] = pd.to_numeric(
    df["rating_count"],
    errors="coerce"
)
```

```python
df[
    [
        "discounted_price",
        "actual_price",
        "discount_percentage",
        "rating",
        "rating_count"
    ]
].dtypes
```

Step 7 — Create useful features

```python
df["amount_saved"] = (
    df["actual_price"] -
    df["discounted_price"]
)
```

```python
df["main_category"] = (
    df["category"]
    .str.split("|")
    .str[0]
)
```

Step 8 — Statistical summary

```python
numeric_columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count",
    "amount_saved"
]

df[numeric_columns].describe()
```

Step 9 — Product category distribution

```python
category_counts = (
    df["main_category"]
    .value_counts()
)

category_counts.head(15)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_counts.head(10).values,
    y=category_counts.head(10).index
)

plt.title("Top 10 Product Categories")
plt.xlabel("Number of Products")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

Step 10 — Price distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="actual_price",
    bins=40,
    kde=True
)

plt.title("Distribution of Actual Product Prices")
plt.xlabel("Actual Price (₹)")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

Step 11 — Discount distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="discount_percentage",
    bins=30,
    kde=True
)

plt.title("Distribution of Product Discounts")
plt.xlabel("Discount Percentage")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

Step 12 — Rating distribution

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="rating"
)

plt.title("Product Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

Step 13 — Rating vs price

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="actual_price",
    y="rating"
)

plt.title("Actual Price vs Product Rating")
plt.xlabel("Actual Price (₹)")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```

Step 14 — Discount vs rating

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="discount_percentage",
    y="rating"
)

plt.title("Discount Percentage vs Product Rating")
plt.xlabel("Discount Percentage")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```

Step 15 — Rating count vs rating

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="rating_count",
    y="rating"
)

plt.title("Rating Count vs Product Rating")
plt.xlabel("Number of Ratings")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```

Step 16 — Category vs discount

```python
category_discount = (
    df.groupby("main_category")
    ["discount_percentage"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

category_discount
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_discount.values,
    y=category_discount.index
)

plt.title("Top Categories by Average Discount")
plt.xlabel("Average Discount (%)")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

Step 17 — Category vs rating

```python
category_rating = (
    df.groupby("main_category")
    ["rating"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

category_rating
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_rating.values,
    y=category_rating.index
)

plt.title("Average Rating by Product Category")
plt.xlabel("Average Rating")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

Step 18 — Correlation analysis

```python
correlation_columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count",
    "amount_saved"
]

correlation = df[
    correlation_columns
].corr()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Product Attributes")

plt.tight_layout()
plt.show()
```

Step 19 — Find strongest relationships

```python
corr_pairs = (
    correlation
    .where(
        np.triu(
            np.ones(correlation.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(
        key=abs,
        ascending=False
    )
)

corr_pairs
```

Step 20 — Top products by rating count

```python
top_reviewed = (
    df[
        [
            "product_name",
            "rating_count"
        ]
    ]
    .sort_values(
        "rating_count",
        ascending=False
    )
    .head(10)
)

top_reviewed
```

```python
plt.figure(figsize=(10, 7))

sns.barplot(
    data=top_reviewed,
    x="rating_count",
    y="product_name"
)

plt.title("Top 10 Most Reviewed Products")
plt.xlabel("Number of Ratings")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```
